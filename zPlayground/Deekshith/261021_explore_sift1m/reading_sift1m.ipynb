The following code helps in understanding how the data is organized in fvecs

```python
import numpy as np
filename = "sift/sift_base.fvecs"
with open(filename, "rb") as f:
    dimension = np.fromfile(f, dtype=np.int32, count=1)
    vector = np.fromfile(f, dtype=np.float32, count=128)

print(f"Dimension = {dimension}")
print(f"First vector = {vector}")
print(f"shape = {vector.shape}")
```

The following code helps understanding the data organization in ivecs

```python
gt_filename = "sift/sift_groundtruth.ivecs"
with open(gt_filename, "rb") as f:
    n_neighbors = np.fromfile(f, dtype=np.int32, count=1)
    indices = np.fromfile(f, dtype=np.int32, count=100)

print(f"k = {n_neighbors}")
print(f"indicies for the first vector = {indices}")
print(f"shape = {indices.shape}")
```
