Here we will revise the sift data organization again and conclude what is what.

```python
import numpy as np
import matplotlib.pyplot as plt
import os
import glob

data_dir = "../../../data/sift/"
lof = glob.glob(os.path.join(data_dir, "*.*"))

# List of files that we obtain on extracting
for file_path in lof:
    file_name = os.path.basename(file_path)
    print(file_name)
```

## Base vector:
The organization of base vector is as follows:

[vector_dim, v_1_1, v_1_1,..., v_1_128,
 vec_dim, v_2_1, v_2_2, ..., v_2_128,
  ...,
 vec_dim, v_1million_1, v_1million_2,..., v_1million_128]

- Here further each vec_dim short form for vector dimension is 32 bit integer.
- v_m_n represents the nth component of the mth vector.
- The above two points mean that 
    - Each vector's data is contained in 1*4 + 128 * 4 bytes = 516  bytes.
- There are 1 million base vectors. And if we index them as from 0 to 9,99,999 then
    - The first vetctor has 0-3rd byte for the vec dim, 4th to 515th byte for the vector data.
    - The data for the vector with index i spans from (516 * i) th byte to (516 * i +515) th byte.
    - The actual vector with index i spans from (516 * i + 4)th byte to (516 * i + 515)th byte.

```python
# Understanding base organization
base_file_name = "sift_base.fvecs"
base_file_path = os.path.join(data_dir, base_file_name)

# Accessing the first vector
with open(base_file_path,"rb") as f:
    dim_each_vector = np.fromfile(f,dtype=np.int32, count=1)
    vector = np.fromfile(f, dtype=np.float32, count=128)
    print(f"dimension = {dim_each_vector}")
    print(f"First vector = {vector}")
```

## Query vector:
The organization of query vector is same as base vector.

[vector_dim, v_1_1, v_1_1,..., v_1_128,
 vec_dim, v_2_1, v_2_2, ..., v_2_128,
  ...,
 vec_dim, v_10000_1, v_10000_2,..., v_10000_128]

The only difference is we have 10000 query vectors.

```python
# Understanding ground truth organization
query_file_name = "sift_query.fvecs"
query_file_path = os.path.join(data_dir, query_file_name)

# Accessing the first vector
with open(query_file_path,"rb") as f:
    dim_each_vector = np.fromfile(f,dtype=np.int32, count=1)
    vector = np.fromfile(f, dtype=np.float32, count=128)
    print(f"dimension = {dim_each_vector}")
    print(f"First vector = {vector}")
```

## Ground truth vector:
The ground truth vector is a book keeper.
For each query vector, the ground truth contains the indices of its 100 nearest base vectors, ordered from nearest to farthest.
For example, suppose the first query has the following nearest neighbors:

[932085 934876 561813 708177 ... 989762]
This means:

Base vector 932085 is the nearest neighbor of query 0.
Base vector 934876 is the second-nearest neighbor.
Base vector 561813 is the third-nearest neighbor.
And so on.

For query q:
- Complete ground truth record \(404q\) to 404q+403
- Neighbor indices \(404q+4\) to 404q+403

```python
# Understanding ground truth organization
gt_file_name = "sift_groundtruth.ivecs"
gt_file_path = os.path.join(data_dir, gt_file_name)

# Accessing the first vector
with open(gt_file_path,"rb") as f:
    dim_each_vector = np.fromfile(f,dtype=np.int32, count=1)
    vector = np.fromfile(f, dtype=np.int32, count=100)
    print(f"dimension = {dim_each_vector}")
    print(f"First vector = {vector}")
```
